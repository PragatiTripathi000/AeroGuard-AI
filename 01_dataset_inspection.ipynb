### Chunk-Based Dataset Audit

**Objective:** To perform an initial audit of the complete dataset by calculating the total number of records, target variable distribution, and missing values without loading the entire dataset into memory.

**Methodology:**

- Used `pd.read_csv()` with `chunksize=100_000` to process the dataset in batches of 100,000 rows.
- Used `len(chunk)` to calculate and accumulate the total number of records.
- Used `value_counts(dropna=False)` to calculate the frequency of each target class (`has_ga`), including missing values.
- Used `isnull().sum()` to calculate missing values in every column.
- Used `.add(fill_value=0)` to accumulate counts across successive chunks.
- Sorted missing-value counts in descending order to identify columns with the highest missingness.

**Results and Observations:**

- Total records processed: **8,969,145**
- Normal landings (`False`): **8,935,481**
- Go-Around events (`True`): **33,664**
- Normal landings: approximately **99.625%**
- Go-Around events: approximately **0.375%**

**Key Findings:**

1. The dataset contains approximately 8.97 million records.
2. The target variable is highly imbalanced, with Go-Around events representing only 0.375% of observations.
3. Missing-value counts were calculated across the complete dataset rather than just a sample.
4. Chunk-based processing reduced memory requirements while allowing us to audit all records.

**Conclusion:**

The initial dataset audit confirmed significant class imbalance. Therefore, during model development, accuracy alone will not be a reliable evaluation metric. Precision, Recall, F1-score, ROC-AUC, PR-AUC, and the confusion matrix will also be considered. The missing-value analysis will guide subsequent data-cleaning and preprocessing decisions.

*Note: This step was performed for data inspection only. No records or original feature values were modified.*

### Missing Value Analysis by Target Class

**Objective:** Investigate whether missing values are distributed differently between normal landings and Go-Around records.

**Why is this important?**

Overall missing-value percentages do not tell us whether missingness is associated with the target class.

Comparing missing values across both classes will help us understand the missing-data patterns before deciding on an imputation strategy.

**Method:** Process the complete dataset in chunks and calculate missing-value percentages separately for each target class.

```python

# Select columns for missing-value analysis.
# Exclude the target column because we are grouping by it.
analysis_cols = [
    col for col in sample_df.columns
    if col != "has_ga"
]

# Initialize a DataFrame to store missing-value counts
# separately for normal landings and Go-Arounds.
missing_by_target = pd.DataFrame(
    0,
    index=[False, True],
    columns=analysis_cols
)

# Store the total number of records in each target class.
rows_by_target = pd.Series(
    0,
    index=[False, True],
    dtype="int64"
)

# Read the complete dataset in chunks.
for chunk in pd.read_csv(
    file_path,
    compression="gzip",
    chunksize=100_000
):

    # Process each target class separately.
    for target_value in [False, True]:

        # Select records belonging to the current target class.
        class_data = chunk[
            chunk["has_ga"] == target_value
        ]

        # Add the number of records in this class.
        rows_by_target[target_value] += len(class_data)

        # Add missing-value counts for each column.
        missing_by_target.loc[target_value] += (
            class_data[analysis_cols].isnull().sum()
        )

# Calculate the missing-value percentage for each class.
missing_percentage = (
    missing_by_target
    .div(rows_by_target, axis=0)
    * 100
)

# Display columns with at least one missing value.
columns_with_missing = missing_percentage.columns[
    (missing_percentage > 0).any()
]

print("MISSING VALUE PERCENTAGE BY TARGET CLASS:")

display(
    missing_percentage[columns_with_missing].round(2)
)
```

2. `press_sea_level_p` has approximately 58–59% missing values in both classes.
3. Aircraft-related features such as `icaoaircrafttype`, `wtc` and `typecode` have comparatively higher missingness among Go-Around records.
4. Weather features such as `wind_speed_knts`, `visibility_m`, `temperature_deg` and `press_p` have very few missing values.

**Preliminary decisions:**

- Investigate the usefulness and availability of `wind_gust_knts` before deciding whether to exclude it.
- Do not perform imputation at this stage.

**Conclusion:**

Missingness varies across features and target classes. High missingness does not automatically mean a feature is useless. Further investigation is required before final feature selection.

### Investigation: Relationship Between Pressure Features

**Objective:** Examine the relationship between `press_p` and `press_sea_level_p`.

**Why is this important?**

Both features represent atmospheric pressure measurements, but they may contain different information because one is adjusted to sea-level conditions.

**Method:** Collect up to 1,000 valid pressure pairs from each dataset chunk and calculate their Pearson correlation.

**Expected outcome:** Understand how strongly the two pressure features are linearly related before deciding whether both should be retained.

```python

# Store valid pressure pairs sampled from each chunk.
pressure_samples = []

# Read only the two pressure columns to reduce memory usage.
for chunk in pd.read_csv(
    file_path,
    compression="gzip",
    usecols=["press_p", "press_sea_level_p"],
    chunksize=100_000
):

    # Keep records where both pressure values are available.
    valid_pressure = chunk.dropna(
        subset=["press_p", "press_sea_level_p"]
    )

    # Randomly select up to 1,000 valid records from this chunk.
    if len(valid_pressure) > 0:
        pressure_samples.append(
            valid_pressure.sample(
                n=min(1000, len(valid_pressure)),
                random_state=42
            )
        )

# Combine the sampled records into one DataFrame.
pressure_df = pd.concat(
    pressure_samples,
    ignore_index=True
)

# Calculate Pearson correlation between the two features.
pressure_correlation = pressure_df[
    ["press_p", "press_sea_level_p"]
].corr()

# Display the correlation matrix.
print("PRESSURE CORRELATION:")
display(pressure_correlation)

# Display basic statistical information.
print("\nPRESSURE SUMMARY:")
display(pressure_df.describe().round(2))
```

### Observations: Pressure Feature Correlation

**Objective:** Investigate the relationship between `press_p` and `press_sea_level_p`.

**Method:** Pearson correlation was calculated using 57,233 valid pressure pairs sampled across dataset chunks.

**Findings:**
- Pearson correlation: 0.994718.
- Both features have very similar statistical distributions.
- `press_p` has approximately 0.02% missing values.
- `press_sea_level_p` has approximately 59% missing values.

**Decision for the initial model:**
- Retain `press_p`.
- Exclude `press_sea_level_p` from the initial feature set.

**Reason:** The two features are extremely strongly correlated, while `press_p` has substantially better data completeness.

This decision can be revisited during model evaluation.

### Categorical Feature Missing Value Handling

**Objective:** To investigate and handle missing values in categorical features without removing any dataset records.

**Methodology:**

- Identified categorical columns containing missing values.
- Replaced missing values in `typecode`, `icaoaircrafttype`, `wtc`, `operator_country`, and `operator_region` with the category `"Unknown"`.
- Handled `has_intersection` separately to preserve its three possible states: True, False, and Unknown.
- Preserved the original weather columns for separate preprocessing.
- Retained all records to avoid unnecessary data loss.

**Expected Outcome:**

All selected categorical columns will have explicit representations for missing information, making them suitable for subsequent categorical encoding and model preparation.

```python

import pandas as pd
import gzip

# Define the input and output file paths
input_file = "data/processed/go_arounds_preprocessing_fixed.csv.gz"
output_file = "data/processed/go_arounds_categorical_checked.csv.gz"

# Categorical columns where missing values will be replaced
categorical_cols = [
    "typecode",
    "icaoaircrafttype",
    "wtc",
    "operator_country",
    "operator_region"
]

# Track whether the first chunk is being processed
first_chunk = True

# Open a new compressed output file
with gzip.open(output_file, "wt", encoding="utf-8", newline="") as output:

    # Read the dataset in chunks of 100,000 rows
    for chunk in pd.read_csv(
        input_file,
        compression="gzip",
        chunksize=100_000
    ):

        # Replace missing values with "Unknown"
        for col in categorical_cols:
            chunk[col] = chunk[col].fillna("Unknown")

        # Preserve True, False and Unknown as separate categories
        chunk["has_intersection"] = (
            chunk["has_intersection"]
            .map({True: "True", False: "False"})
            .fillna("Unknown")
        )

        # Write the processed chunk to the compressed output file
        chunk.to_csv(
            output,
            index=False,
            header=first_chunk
        )

        # Only the first chunk should write column headers
        first_chunk = False

print("CATEGORICAL PREPROCESSING COMPLETED")
print("OUTPUT FILE:", output_file)
```

```python

import pandas as pd
import gzip
import os
from collections import Counter

# --------------------------------------------------
# STEP 1: DEFINE INPUT AND OUTPUT PATHS
# --------------------------------------------------

input_file = "data/processed/go_arounds_runway_checked.csv.gz"

# Create a folder for the three datasets
output_folder = "data/processed/splits"
os.makedirs(output_folder, exist_ok=True)

output_files = {
    "train": output_folder + "/train.csv.gz",
    "validation": output_folder + "/validation.csv.gz",
    "test": output_folder + "/test.csv.gz"
}

# --------------------------------------------------
# STEP 2: DEFINE TEMPORAL SPLIT BOUNDARIES
# --------------------------------------------------

# Training: January to September 2019
# Validation: October 2019
# Testing: November to December 2019

train_end = pd.Timestamp("2019-10-01", tz="UTC")
validation_end = pd.Timestamp("2019-11-01", tz="UTC")
test_end = pd.Timestamp("2020-01-01", tz="UTC")

# --------------------------------------------------
# STEP 3: INITIALIZE COUNTERS
# --------------------------------------------------

row_counts = Counter()

target_counts = {
    "train": Counter(),
    "validation": Counter(),
    "test": Counter()
}

first_chunk = {
    "train": True,
    "validation": True,
    "test": True
}

# --------------------------------------------------
# STEP 4: OPEN OUTPUT FILES
# --------------------------------------------------

with gzip.open(output_files["train"], "wt",
               encoding="utf-8", newline="") as train_output, \
     gzip.open(output_files["validation"], "wt",
               encoding="utf-8", newline="") as validation_output, \
     gzip.open(output_files["test"], "wt",
               encoding="utf-8", newline="") as test_output:

    output_handles = {
        "train": train_output,
        "validation": validation_output,
        "test": test_output
    }

    # --------------------------------------------------
    # STEP 5: READ THE DATASET IN CHUNKS
    # --------------------------------------------------

    for chunk in pd.read_csv(
        input_file,
        compression="gzip",
        chunksize=100_000
    ):

        # Convert timestamp into UTC datetime format
        timestamps = pd.to_datetime(
            chunk["time"],
            utc=True,
            errors="coerce"
        )

        # Stop if any timestamp is invalid
        if timestamps.isna().any():
            raise ValueError("Invalid timestamps detected!")

        # --------------------------------------------------
        # STEP 6: SEPARATE RECORDS BY DATE
        # --------------------------------------------------

        train_mask = timestamps < train_end

        validation_mask = (
            (timestamps >= train_end) &
            (timestamps < validation_end)
        )

        test_mask = (
            (timestamps >= validation_end) &
            (timestamps < test_end)
        )

        masks = {
            "train": train_mask,
            "validation": validation_mask,
            "test": test_mask
        }

        # --------------------------------------------------
        # STEP 7: SAVE EACH SPLIT
        # --------------------------------------------------

        for split_name, mask in masks.items():

            split_chunk = chunk.loc[mask]

            if len(split_chunk) == 0:
                continue

            # Save records to the corresponding file
            split_chunk.to_csv(
                output_handles[split_name],
                index=False,
                header=first_chunk[split_name]
            )

            first_chunk[split_name] = False

            # Count total records
            row_counts[split_name] += len(split_chunk)

            # Count normal and Go-Around records
            target_counts[split_name].update(
                split_chunk["has_ga"].value_counts().to_dict()
            )

# --------------------------------------------------
# STEP 8: DISPLAY SPLIT RESULTS
# --------------------------------------------------

print("TEMPORAL DATA SPLITTING COMPLETED")

for split_name in ["train", "validation", "test"]:

    print("\n" + "=" * 45)
    print("DATASET:", split_name.upper())
    print("TOTAL RECORDS:", row_counts[split_name])
    print("NORMAL LANDINGS:", target_counts[split_name][False])
    print("GO-AROUNDS:", target_counts[split_name][True])

    print("OUTPUT FILE:", output_files[split_name])

print("\nTOTAL RECORDS:", sum(row_counts.values()))
```

```python

import pandas as pd
import numpy as np
import gzip
import joblib

from collections import Counter
from sklearn.preprocessing import OneHotEncoder

# Training dataset
train_file = "data/processed/splits/train_standardized.csv.gz"

# Features for One-Hot Encoding
onehot_cols = [
    "wtc",
    "has_intersection",
    "airport_region",
    "operator_region",
    "icaoaircrafttype",
    "airport_country"
]

# Features for Frequency Encoding
frequency_cols = [
    "airport",
    "runway",
    "typecode",
    "operator_country"
]

# Combine all categorical features
categorical_cols = onehot_cols + frequency_cols

# Store category counts from training data
category_counts = {
    col: Counter() for col in categorical_cols
}

total_records = 0

# Read training data in chunks
for chunk in pd.read_csv(
    train_file,
    compression="gzip",
    chunksize=100_000,
    usecols=categorical_cols,
    dtype={col: "string" for col in categorical_cols}
):

    total_records += len(chunk)

    # Count every category in each feature
    for col in categorical_cols:
        category_counts[col].update(
            chunk[col].value_counts().to_dict()
        )

print("TRAINING CATEGORY COUNTS COLLECTED.")
print("Total training records:", total_records)

# Prepare category lists for One-Hot Encoding
onehot_categories = [
    sorted(category_counts[col].keys())
    for col in onehot_cols
]

# Create One-Hot Encoder
encoder = OneHotEncoder(
    categories=onehot_categories,
    handle_unknown="ignore",
    sparse_output=True,
    dtype=np.float32
)

# Fit using one example per feature.
# The complete category lists were learned from training data.
fit_example = pd.DataFrame([
    [categories[0] for categories in onehot_categories]
], columns=onehot_cols)

encoder.fit(fit_example)

# Create frequency mappings using training frequencies
frequency_maps = {}

for col in frequency_cols:

    frequency_maps[col] = {
        category: count / total_records
        for category, count in category_counts[col].items()
    }

# Save the learned encoding information
encoding_artifacts = {
    "onehot_encoder": encoder,
    "frequency_maps": frequency_maps,
    "onehot_cols": onehot_cols,
    "frequency_cols": frequency_cols
}

joblib.dump(
    encoding_artifacts,
    "models/categorical_encoder.joblib"
)

# Display encoding summary
print("\nONE-HOT ENCODING SUMMARY")

for col in onehot_cols:
    print(col, ":", len(category_counts[col]), "categories")

print("\nTOTAL ONE-HOT FEATURES:", len(encoder.get_feature_names_out()))

print("\nFREQUENCY ENCODING SUMMARY")

for col in frequency_cols:
    print(col, ":", len(frequency_maps[col]), "categories")

print("\nENCODER SAVED SUCCESSFULLY.")
```

```python

# Extract exact two-character METAR codes.
def extract_weather_tokens(codes):

    tokens = set()

    for code in codes:

        # Ignore missing and empty values.
        if code is None or not isinstance(code, str):
            continue

        code = code.strip()

        # Ignore blank and slash-based placeholders.
        if not code or "/" in code:
            continue

        # Split combined codes into two-character tokens.
        if len(code) % 2 == 0:
            for i in range(0, len(code), 2):
                tokens.add(code[i:i+2])

    return tokens


# Define weather feature groups.
weather_code_groups = {
    "has_rain": ["RA"],
    "has_drizzle": ["DZ"],
    "has_snow": ["SN", "SG"],
    "has_thunderstorm": ["TS"],
    "has_showers": ["SH"],
    "has_hail": ["GR", "GS"],
    "has_ice_pellets": ["PL"],
    "has_fog": ["FG"],
    "has_mist": ["BR"],
    "has_haze": ["HZ"],
    "has_dust": ["DU", "DS"],
    "has_sand": ["SA", "SS"],
    "has_smoke": ["FU"],
    "has_squalls": ["SQ"],
    "has_freezing": ["FZ"],
}

# Test combined and individual codes.
test_weather_codes = [
    ["RA"], ["RASN"], ["RADZ"], ["GSRA"],
    ["FZBC"], ["FZPR"], ["SNPLRA"],
    ["BR"], ["FG"], ["SQ"],
    ["//"], [None], [""]
]

for codes in test_weather_codes:

    tokens = extract_weather_tokens(codes)

    print("\nINPUT:", codes)
    print("EXTRACTED TOKENS:", tokens)

    for feature, patterns in weather_code_groups.items():

        detected = bool(tokens.intersection(patterns))

        if detected:
            print(feature, ":", detected)
```

```python
 
import pandas as pd
import ast
import os

# --------------------------------------------------
# 1. Input and output paths
# --------------------------------------------------

input_files = {
    "train": "data/processed/encoded/train_encoded.csv.gz",
    "validation": "data/processed/encoded/validation_encoded.csv.gz",
    "test": "data/processed/encoded/test_encoded.csv.gz"
}

output_folder = "data/processed/weather"
os.makedirs(output_folder, exist_ok=True)

weather_columns = [
    "weather_intensity",
    "weather_precipitation",
    "weather_desc",
    "weather_obscuration",
    "weather_other"
]


# --------------------------------------------------
# 2. Define weather feature mappings
# --------------------------------------------------

weather_feature_groups = {
    "has_rain": ("weather_desc", ["RA"]),
    "has_drizzle": ("weather_desc", ["DZ"]),
    "has_snow": ("weather_desc", ["SN", "SG"]),
    "has_hail": ("weather_desc", ["GR", "GS"]),
    "has_ice_pellets": ("weather_desc", ["PL"]),

    "has_showers": ("weather_precipitation", ["SH"]),
    "has_thunderstorm": ("weather_precipitation", ["TS"]),
    "has_freezing": ("weather_precipitation", ["FZ"]),

    "has_mist": ("weather_obscuration", ["BR"]),
    "has_fog": ("weather_obscuration", ["FG"]),
    "has_haze": ("weather_obscuration", ["HZ"]),
    "has_dust": ("weather_obscuration", ["DU"]),
    "has_smoke": ("weather_obscuration", ["FU"]),
    "has_sand": ("weather_obscuration", ["SA"]),

    "has_squalls": ("weather_other", ["SQ"]),
    "has_duststorm": ("weather_other", ["DS"]),
    "has_funnel_cloud": ("weather_other", ["FC"]),
    "has_sandstorm": ("weather_other", ["SS"])
}


# --------------------------------------------------
# 3. Function to parse one unique weather pattern
# --------------------------------------------------

def parse_weather_pattern(pattern, split_combined=False):

    if not isinstance(pattern, str):
        return frozenset()

    try:
        values = ast.literal_eval(pattern)
    except (ValueError, SyntaxError):
        return frozenset()

    tokens = set()

    for value in values:

        if not isinstance(value, str):
            continue

        value = value.strip()

        if not value or "/" in value:
            continue

        if split_combined:
            if len(value) % 2 == 0:
                for i in range(0, len(value), 2):
                    tokens.add(value[i:i+2])
        else:
            tokens.add(value)

    return frozenset(tokens)


# --------------------------------------------------
# 4. Prepare reusable lookup dictionaries
# --------------------------------------------------

# Each unique serialized pattern will be parsed only once.
token_cache = {
    column: {} for column in weather_columns
}

# Each feature gets a lookup dictionary.
feature_lookup = {
    feature: {} for feature in weather_feature_groups
}

intensity_lookup = {
    "has_heavy_intensity": {},
    "has_light_intensity": {},
    "has_vicinity": {}
}

# Track row position separately for later alignment.
row_start = 0


# --------------------------------------------------
# 5. Process train, validation and test
# --------------------------------------------------

for split_name, input_file in input_files.items():

    output_file = (
        f"{output_folder}/{split_name}_weather_features.csv.gz"
    )

    # Remove incomplete output from the earlier attempt.
    if os.path.exists(output_file):
        os.remove(output_file)

    row_start = 0

    print("\nPROCESSING:", split_name.upper())

    for chunk in pd.read_csv(
        input_file,
        compression="gzip",
        chunksize=100_000,
        usecols=weather_columns
    ):

        # ------------------------------------------
        # Parse only previously unseen patterns.
        # ------------------------------------------

        for column in weather_columns:

            split_combined = column != "weather_intensity"

            for pattern in chunk[column].dropna().unique():

                if pattern not in token_cache[column]:

                    token_cache[column][pattern] = (
                        parse_weather_pattern(
                            pattern,
                            split_combined
                        )
                    )

        # ------------------------------------------
        # Create weather condition indicators.
        # ------------------------------------------

        output = pd.DataFrame(index=chunk.index)

        for feature, (column, codes) in weather_feature_groups.items():

            # Build lookup values for known patterns.
            lookup = feature_lookup[feature]

            for pattern, tokens in token_cache[column].items():

                if pattern not in lookup:
                    lookup[pattern] = bool(
                        tokens.intersection(codes)
                    )

            output[feature] = (
                chunk[column]
                .map(lookup)
                .fillna(False)
                .astype("int8")
            )

        # ------------------------------------------
        # Create intensity indicators.
        # ------------------------------------------

        intensity_tokens = token_cache["weather_intensity"]

        for pattern, tokens in intensity_tokens.items():

            if pattern not in intensity_lookup["has_heavy_intensity"]:
                intensity_lookup["has_heavy_intensity"][pattern] = (
                    "+" in tokens
                )

                intensity_lookup["has_light_intensity"][pattern] = (
                    "-" in tokens
                )

                intensity_lookup["has_vicinity"][pattern] = (
                    "VC" in tokens
                )

        for feature, lookup in intensity_lookup.items():

            output[feature] = (
                chunk["weather_intensity"]
                .map(lookup)
                .fillna(False)
                .astype("int8")
            )

        # ------------------------------------------
        # Indicate whether a recognized condition
        # or intensity code was detected.
        # ------------------------------------------

        feature_columns = list(weather_feature_groups) + list(
            intensity_lookup
        )

        output["weather_codes_present"] = (
            output[feature_columns].any(axis=1).astype("int8")
        )

        # Preserve original row order using row_index.
        output.insert(
            0,
            "row_index",
            range(row_start, row_start + len(chunk))
        )

        # ------------------------------------------
        # Save only compact engineered features.
        # ------------------------------------------

        output.to_csv(
            output_file,
            compression="gzip",
            mode="a",
            header=(row_start == 0),
            index=False
        )

        row_start += len(chunk)

        print("ROWS PROCESSED:", row_start)

    print("COMPLETED:", split_name.upper())
    print("TOTAL ROWS:", row_start)
    print("OUTPUT:", output_file)

print("\nALL WEATHER FEATURES CREATED SUCCESSFULLY.")
```

```python

import pandas as pd
import os

# --------------------------------------------------
# 1. Define input and output files
# --------------------------------------------------

numerical_files = {
    "train": "data/processed/numerical/train_numerical_imputed.csv.gz",
    "validation": "data/processed/numerical/validation_numerical_imputed.csv.gz",
    "test": "data/processed/numerical/test_numerical_imputed.csv.gz"
}

weather_files = {
    "train": "data/processed/weather/train_weather_features.csv.gz",
    "validation": "data/processed/weather/validation_weather_features.csv.gz",
    "test": "data/processed/weather/test_weather_features.csv.gz"
}

output_folder = "data/processed/model_ready"
os.makedirs(output_folder, exist_ok=True)

# Original weather text columns to remove.
raw_weather_columns = [
    "weather_intensity",
    "weather_precipitation",
    "weather_desc",
    "weather_obscuration",
    "weather_other"
]


# --------------------------------------------------
# 2. Combine each dataset in matching chunks
# --------------------------------------------------

for split_name in numerical_files:

    numerical_reader = pd.read_csv(
        numerical_files[split_name],
        compression="gzip",
        chunksize=100_000
    )

    weather_reader = pd.read_csv(
        weather_files[split_name],
        compression="gzip",
        chunksize=100_000
    )

    output_file = (
        f"{output_folder}/{split_name}_model_ready.csv.gz"
    )

    if os.path.exists(output_file):
        os.remove(output_file)

    total_rows = 0
    first_chunk = True

    print("\nPROCESSING:", split_name.upper())

    for numerical_chunk, weather_chunk in zip(
        numerical_reader,
        weather_reader
    ):

        # ------------------------------------------
        # Verify that both chunks contain the same
        # number of records and matching row indexes.
        # ------------------------------------------

        if len(numerical_chunk) != len(weather_chunk):
            raise ValueError("Chunk row counts do not match.")

        expected_indexes = range(
            total_rows,
            total_rows + len(numerical_chunk)
        )

        if weather_chunk["row_index"].tolist() != list(
            expected_indexes
        ):
            raise ValueError("Weather row alignment failed.")

        # ------------------------------------------
        # Remove the original weather text columns.
        # ------------------------------------------

        numerical_chunk = numerical_chunk.drop(
            columns=raw_weather_columns
        )

        # ------------------------------------------
        # Add engineered weather features.
        # ------------------------------------------

        weather_features = weather_chunk.drop(
            columns=["row_index", "weather_codes_present"]
        )

        combined_chunk = pd.concat(
            [
                numerical_chunk.reset_index(drop=True),
                weather_features.reset_index(drop=True),
                weather_chunk["weather_codes_present"]
                .rename("recognized_weather_condition_present")
                .reset_index(drop=True)
            ],
            axis=1
        )

        # ------------------------------------------
        # Save the combined dataset.
        # ------------------------------------------

        combined_chunk.to_csv(
            output_file,
            compression="gzip",
            mode="a",
            header=first_chunk,
            index=False
        )

        first_chunk = False
        total_rows += len(combined_chunk)

        print("ROWS COMBINED:", total_rows)

    # Ensure neither input contains extra chunks.
    if next(numerical_reader, None) is not None:
        raise ValueError("Numerical file has extra rows.")

    if next(weather_reader, None) is not None:
        raise ValueError("Weather file has extra rows.")

    print("COMPLETED:", split_name.upper())
    print("TOTAL ROWS:", total_rows)
    print("OUTPUT:", output_file)

print("\nALL THREE MODEL-READY DATASETS CREATED.")
```

**Objective:** Investigate whether the absence of recorded weather information can provide useful information for Go-Around prediction.

- `1`: All five weather fields are empty.
- `0`: At least one weather field contains recorded content.
